```python
from sqlalchemy import create_engine
from app.modelos.modelos import Base
from app.servicios.geocodeService import GeocodeService
from app.servicios.pronosticoService import PronosticoService
from app.use_cases.fill_cities_database import FillCitiesDatabase
from app.use_cases.fill_pronosticos_database import FillPronosticosDatabase

# Configuración de la base de datos
DATABASE_URL = "postgresql+psycopg2://postgres:123456@localhost:5434/Weather"
engine = create_engine(DATABASE_URL)
Base.metadata.create_all(engine)
print("✅ Tablas creadas si no existían previamente.")

# Instanciar servicios
geocode_service = GeocodeService()
pronostico_service = PronosticoService()

# Ejecutar casos de uso
cities = ["Barcelona", "Madrid", "Valencia", "Sevilla", "Bilbao"]

# Llenar ciudades
fill_cities = FillCitiesDatabase(geocode_service)
if fill_cities.execute(cities):
    print("✅ Ciudades insertadas correctamente.")
else:
    print("❌ Error al insertar ciudades.")

# Llenar pronósticos
fill_pronosticos = FillPronosticosDatabase(geocode_service, pronostico_service)
if fill_pronosticos.execute(cities):
    print("✅ Pronósticos insertados correctamente.")
else:
    print("❌ Error al insertar pronósticos.")
```
